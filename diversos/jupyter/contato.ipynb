#### Inserir

```python

from database.conexao import conectar
from models.contato import Contato
from peewee import IntegrityError

dados = {"nm_contato": "Diego Alves",
        "email": "diego@gmail.com.br",
        "celular": "11 98741-79452",
        "dt_nascimento": "1985-09-02"}

with conectar():
    try:
        crud = Contato()
        contato = crud.inserir(**dados)
        print(f"Contato: {contato.nm_contato} inserido.")
    except TypeError as erro:
        print(f"Erro de tipo: {erro}")
    except IntegrityError as erro:
         print(f"Erro do PostgreSQL: {erro}")
        
```
